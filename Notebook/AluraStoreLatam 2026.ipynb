### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
ingreso_tienda1 = tienda["Precio"].sum()
ingreso_tienda2 = tienda2["Precio"].sum()
ingreso_tienda3 = tienda3["Precio"].sum()
ingreso_tienda4 = tienda4["Precio"].sum()

print("Ingreso Tienda 1:", ingreso_tienda1)
print("Ingreso Tienda 2:", ingreso_tienda2)
print("Ingreso Tienda 3:", ingreso_tienda3)
print("Ingreso Tienda 4:", ingreso_tienda4)
```

```python
import matplotlib.pyplot as plt

ingresos = [ingreso_tienda1, ingreso_tienda2, ingreso_tienda3, ingreso_tienda4]
nombres_tiendas = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]

plt.figure(figsize=(8,5))
plt.bar(nombres_tiendas, ingresos, color=["#4e79a7", "#59a14f", "#f28e2b", "#e15759"])
plt.title("Ingresos totales por tienda")
plt.xlabel("Tiendas")
plt.ylabel("Ingresos")
plt.show()
```

# Informe Final – Análisis de desempeño de las tiendas Alura Store

## Introducción

El presente análisis tiene como objetivo evaluar el desempeño de las cuatro tiendas de la cadena Alura Store con el fin de recomendar cuál de ellas debería vender el Sr. Juan para iniciar un nuevo emprendimiento. Para ello, se analizaron diversos indicadores relevantes como los ingresos totales de cada tienda, las categorías de productos más y menos vendidas, las calificaciones promedio de los clientes, los productos con mayor y menor volumen de ventas y el costo promedio de envío.

A través del análisis de estos datos y de la generación de visualizaciones, es posible identificar patrones de rendimiento entre las tiendas y determinar cuál presenta el menor desempeño general.

---

## Análisis de ingresos de las tiendas

El análisis de facturación permitió comparar los ingresos generados por cada una de las tiendas. Los resultados muestran que las tiendas presentan niveles de ingresos relativamente similares, sin embargo existen diferencias claras en el rendimiento general.

Las tiendas 1, 2 y 3 presentan niveles de ingresos superiores en comparación con la tienda 4. Esto indica que la tienda 4 tiene un menor volumen de ventas y, por lo tanto, un desempeño financiero más bajo en comparación con las demás tiendas.

Este indicador es uno de los factores más importantes para la toma de decisiones, ya que refleja directamente la capacidad de generación de ingresos de cada tienda.

---

## Ventas por categoría de productos

El análisis de ventas por categoría permitió identificar cuáles son los tipos de productos más demandados por los clientes. Se observó que categorías como **Electrónicos y Muebles** tienden a concentrar una mayor cantidad de ventas en comparación con otras categorías.

Por otro lado, categorías como **libros u otros productos especializados** presentan un menor volumen de ventas.

Este análisis es importante porque permite comprender el perfil de consumo de los clientes y la relevancia de cada categoría dentro del negocio. Las tiendas que presentan un mejor equilibrio entre categorías suelen tener un desempeño más estable.

---

## Calificación promedio de las tiendas

La satisfacción del cliente se evaluó a través de la calificación promedio de las compras. En general, las tiendas presentan valores de calificación similares, lo que indica un nivel aceptable de satisfacción del cliente en toda la cadena.

Sin embargo, algunas tiendas presentan ligeras variaciones en su promedio de calificaciones. Las tiendas con mejores calificaciones reflejan una experiencia de compra más satisfactoria para los clientes, lo que puede contribuir a la fidelización y a futuras ventas.

---

## Productos más y menos vendidos

El análisis de los productos más vendidos permitió identificar cuáles son los artículos con mayor rotación dentro de cada tienda. Los productos más vendidos suelen concentrarse en categorías populares como electrónicos, muebles y artículos de uso cotidiano.

Por otro lado, los productos menos vendidos corresponden generalmente a artículos más específicos o de menor demanda.

Este análisis es útil para comprender la dinámica de ventas de cada tienda y detectar posibles oportunidades de optimización en el inventario.

---

## Costo promedio de envío por tienda

El análisis del costo promedio de envío permitió identificar diferencias entre las tiendas en relación con los gastos asociados a la logística de entrega.

En general, las tiendas presentan costos de envío relativamente similares, aunque algunas pueden presentar valores ligeramente más altos. Un costo de envío elevado puede afectar la decisión de compra de los clientes y disminuir la competitividad de la tienda.

Por lo tanto, este indicador también es relevante al evaluar el desempeño general del negocio.

---

## Conclusión y recomendación

Con base en los análisis realizados y en la comparación de los diferentes indicadores evaluados, se recomienda que **el Sr. Juan venda la tienda 4**.

Esta recomendación se fundamenta principalmente en que la tienda 4 presenta un **menor desempeño general en términos de ingresos** en comparación con las otras tiendas. Aunque los demás indicadores como calificaciones o categorías de productos pueden ser similares entre tiendas, el nivel de facturación es un factor clave para evaluar la eficiencia del negocio.

Las tiendas 1, 2 y 3 muestran un desempeño más sólido en términos de generación de ingresos y volumen de ventas, por lo que representan activos más valiosos para mantener dentro del negocio.

En consecuencia, vender la tienda 4 permitiría al Sr. Juan concentrar sus recursos en las tiendas con mejor rendimiento y utilizar el capital obtenido para iniciar su nuevo emprendimiento.
